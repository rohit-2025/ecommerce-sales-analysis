# E-Commerce Sales Analysis - Complete Analysis and Visualizations

This notebook contains complete sales analysis with visualizations, charts, and business insights.

```python3
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Set styling
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (14, 8)
plt.rcParams['font.size'] = 10

# Load the cleaned data
df = pd.read_csv('data/processed/sales_data_cleaned.csv')
df['Date'] = pd.to_datetime(df['Date'])

print("Data loaded successfully")
print(f"Dataset shape: {df.shape}")
print(f"Date range: {df['Date'].min()} to {df['Date'].max()}")
```

## Executive Summary - Key Metrics

```python3
# Calculate all key metrics
total_revenue = df['Revenue'].sum()
total_orders = len(df)
avg_order_value = df['Revenue'].mean()
total_items = df['Quantity'].sum()
avg_rating = df['Rating'].mean()
unique_customers = df['Customer_ID'].nunique()
total_profit = df['Profit'].sum()
avg_profit_margin = df['Profit_Margin'].mean()

print("\n" + "="*60)
print("EXECUTIVE SUMMARY - KEY PERFORMANCE INDICATORS")
print("="*60)
print(f"Total Revenue:              Rs {total_revenue:>15,.2f}")
print(f"Total Orders:               {total_orders:>20}")
print(f"Average Order Value:        Rs {avg_order_value:>15,.2f}")
print(f"Total Items Sold:           {total_items:>20} units")
print(f"Total Profit:               Rs {total_profit:>15,.2f}")
print(f"Average Profit Margin:      {avg_profit_margin:>18.2f}%")
print(f"Unique Customers:           {unique_customers:>20}")
print(f"Average Customer Rating:    {avg_rating:>18.2f}/5.0")
print("="*60)
```

## Sales Analysis by Category

```python3
# Category analysis
category_analysis = df.groupby('Category').agg({
    'Revenue': 'sum',
    'Order_ID': 'count',
    'Quantity': 'sum',
    'Rating': 'mean',
    'Profit': 'sum'
}).round(2)

category_analysis.columns = ['Total_Revenue', 'Orders', 'Units_Sold', 'Avg_Rating', 'Total_Profit']
category_analysis = category_analysis.sort_values('Total_Revenue', ascending=False)

print("\nSALES BY CATEGORY:")
print(category_analysis)
print(f"\nTop Category: {category_analysis.index[0]} with Rs {category_analysis.iloc[0]['Total_Revenue']:,.2f} revenue")
```

```python3
# Visualize category analysis
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('Sales Analysis by Category', fontsize=16, fontweight='bold', y=1.00)

# Revenue by category
category_revenue = df.groupby('Category')['Revenue'].sum().sort_values(ascending=False)
axes[0, 0].bar(category_revenue.index, category_revenue.values, color='steelblue', edgecolor='black')
axes[0, 0].set_title('Total Revenue by Category', fontweight='bold', fontsize=12)
axes[0, 0].set_ylabel('Revenue (Rs)')
axes[0, 0].tick_params(axis='x', rotation=45)
for i, v in enumerate(category_revenue.values):
    axes[0, 0].text(i, v, f'Rs {v:,.0f}', ha='center', va='bottom', fontsize=9)

# Orders by category
category_orders = df['Category'].value_counts()
axes[0, 1].bar(category_orders.index, category_orders.values, color='coral', edgecolor='black')
axes[0, 1].set_title('Number of Orders by Category', fontweight='bold', fontsize=12)
axes[0, 1].set_ylabel('Number of Orders')
axes[0, 1].tick_params(axis='x', rotation=45)
for i, v in enumerate(category_orders.values):
    axes[0, 1].text(i, v, f'{v}', ha='center', va='bottom', fontsize=9)

# Revenue distribution pie chart
axes[1, 0].pie(category_revenue.values, labels=category_revenue.index, autopct='%1.1f%%', 
               colors=['steelblue', 'coral', 'lightgreen', 'gold', 'salmon', 'lightblue'],
               startangle=90)
axes[1, 0].set_title('Revenue Distribution by Category', fontweight='bold', fontsize=12)

# Average rating by category
category_rating = df.groupby('Category')['Rating'].mean().sort_values(ascending=False)
axes[1, 1].barh(category_rating.index, category_rating.values, color='lightgreen', edgecolor='black')
axes[1, 1].set_title('Average Customer Rating by Category', fontweight='bold', fontsize=12)
axes[1, 1].set_xlabel('Average Rating')
axes[1, 1].set_xlim(0, 5)
for i, v in enumerate(category_rating.values):
    axes[1, 1].text(v, i, f'{v:.2f}', ha='left', va='center', fontsize=9)

plt.tight_layout()
plt.show()
print("Category analysis visualization complete")
```

## Regional Performance Analysis

```python3
# Regional analysis
region_analysis = df.groupby('Region').agg({
    'Revenue': 'sum',
    'Order_ID': 'count',
    'Customer_ID': 'nunique',
    'Profit': 'sum',
    'Rating': 'mean'
}).round(2)

region_analysis.columns = ['Total_Revenue', 'Orders', 'Customers', 'Total_Profit', 'Avg_Rating']
region_analysis = region_analysis.sort_values('Total_Revenue', ascending=False)

print("\nSALES BY REGION:")
print(region_analysis)
print(f"\nTop Region: {region_analysis.index[0]} with Rs {region_analysis.iloc[0]['Total_Revenue']:,.2f} revenue")
```

```python3
# Visualize regional analysis
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('Regional Performance Analysis', fontsize=16, fontweight='bold', y=1.00)

# Revenue by region
region_revenue = df.groupby('Region')['Revenue'].sum().sort_values(ascending=False)
axes[0, 0].bar(region_revenue.index, region_revenue.values, color='steelblue', edgecolor='black')
axes[0, 0].set_title('Total Revenue by Region', fontweight='bold', fontsize=12)
axes[0, 0].set_ylabel('Revenue (Rs)')
for i, v in enumerate(region_revenue.values):
    axes[0, 0].text(i, v, f'Rs {v:,.0f}', ha='center', va='bottom', fontsize=9)

# Orders by region
region_orders = df['Region'].value_counts()
axes[0, 1].bar(region_orders.index, region_orders.values, color='coral', edgecolor='black')
axes[0, 1].set_title('Number of Orders by Region', fontweight='bold', fontsize=12)
axes[0, 1].set_ylabel('Number of Orders')
for i, v in enumerate(region_orders.values):
    axes[0, 1].text(i, v, f'{v}', ha='center', va='bottom', fontsize=9)

# Revenue distribution pie chart
axes[1, 0].pie(region_revenue.values, labels=region_revenue.index, autopct='%1.1f%%',
               colors=['steelblue', 'coral', 'lightgreen', 'gold', 'salmon'],
               startangle=90)
axes[1, 0].set_title('Revenue Distribution by Region', fontweight='bold', fontsize=12)

# Customers by region
region_customers = df.groupby('Region')['Customer_ID'].nunique().sort_values(ascending=False)
axes[1, 1].barh(region_customers.index, region_customers.values, color='lightgreen', edgecolor='black')
axes[1, 1].set_title('Unique Customers by Region', fontweight='bold', fontsize=12)
axes[1, 1].set_xlabel('Number of Customers')
for i, v in enumerate(region_customers.values):
    axes[1, 1].text(v, i, f'{v}', ha='left', va='center', fontsize=9)

plt.tight_layout()
plt.show()
print("Regional analysis visualization complete")
```

## Top Products Analysis

```python3
# Top 10 products
top_products = df.groupby('Product').agg({
    'Revenue': 'sum',
    'Quantity': 'sum',
    'Order_ID': 'count',
    'Rating': 'mean',
    'Profit': 'sum'
}).round(2)

top_products.columns = ['Total_Revenue', 'Units_Sold', 'Orders', 'Avg_Rating', 'Total_Profit']
top_products = top_products.sort_values('Total_Revenue', ascending=False).head(10)

print("\nTOP 10 PRODUCTS BY REVENUE:")
print(top_products)
```

```python3
# Visualize top products
fig, axes = plt.subplots(1, 2, figsize=(16, 8))
fig.suptitle('Top 10 Products Analysis', fontsize=16, fontweight='bold')

# Top products by revenue
top_products_sorted = df.groupby('Product')['Revenue'].sum().sort_values(ascending=True).tail(10)
axes[0].barh(range(len(top_products_sorted)), top_products_sorted.values, color='steelblue', edgecolor='black')
axes[0].set_yticks(range(len(top_products_sorted)))
axes[0].set_yticklabels(top_products_sorted.index, fontsize=9)
axes[0].set_title('Top 10 Products by Revenue', fontweight='bold', fontsize=12)
axes[0].set_xlabel('Revenue (Rs)')
for i, v in enumerate(top_products_sorted.values):
    axes[0].text(v, i, f' Rs {v:,.0f}', ha='left', va='center', fontsize=8)

# Top products by units sold
top_units = df.groupby('Product')['Quantity'].sum().sort_values(ascending=True).tail(10)
axes[1].barh(range(len(top_units)), top_units.values, color='coral', edgecolor='black')
axes[1].set_yticks(range(len(top_units)))
axes[1].set_yticklabels(top_units.index, fontsize=9)
axes[1].set_title('Top 10 Products by Units Sold', fontweight='bold', fontsize=12)
axes[1].set_xlabel('Units Sold')
for i, v in enumerate(top_units.values):
    axes[1].text(v, i, f' {v} units', ha='left', va='center', fontsize=8)

plt.tight_layout()
plt.show()
print("Top products visualization complete")
```

## Monthly Sales Trends

```python3
# Monthly sales trend
monthly_sales = df.groupby(df['Date'].dt.to_period('M')).agg({
    'Revenue': 'sum',
    'Order_ID': 'count',
    'Quantity': 'sum',
    'Profit': 'sum'
}).round(2)

monthly_sales.columns = ['Total_Revenue', 'Orders', 'Units_Sold', 'Total_Profit']

print("\nMONTHLY SALES TREND:")
print(monthly_sales)
```

```python3
# Visualize monthly trends
fig, axes = plt.subplots(2, 1, figsize=(14, 10))
fig.suptitle('Monthly Sales Trends', fontsize=16, fontweight='bold')

months = [str(period) for period in monthly_sales.index]
x_pos = range(len(months))

# Monthly revenue trend
axes[0].plot(x_pos, monthly_sales['Total_Revenue'].values, marker='o', linewidth=2.5, 
            markersize=10, color='steelblue', label='Revenue')
axes[0].bar(x_pos, monthly_sales['Total_Revenue'].values, alpha=0.3, color='steelblue')
axes[0].set_title('Monthly Revenue Trend', fontweight='bold', fontsize=12)
axes[0].set_ylabel('Revenue (Rs)')
axes[0].set_xticks(x_pos)
axes[0].set_xticklabels(months, rotation=45)
axes[0].grid(True, alpha=0.3)
for i, v in enumerate(monthly_sales['Total_Revenue'].values):
    axes[0].text(i, v, f'Rs {v:,.0f}', ha='center', va='bottom', fontsize=9)

# Monthly orders trend
axes[1].plot(x_pos, monthly_sales['Orders'].values, marker='s', linewidth=2.5,
            markersize=10, color='coral', label='Orders')
axes[1].bar(x_pos, monthly_sales['Orders'].values, alpha=0.3, color='coral')
axes[1].set_title('Monthly Orders Trend', fontweight='bold', fontsize=12)
axes[1].set_ylabel('Number of Orders')
axes[1].set_xlabel('Month')
axes[1].set_xticks(x_pos)
axes[1].set_xticklabels(months, rotation=45)
axes[1].grid(True, alpha=0.3)
for i, v in enumerate(monthly_sales['Orders'].values):
    axes[1].text(i, v, f'{int(v)}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
print("Monthly trends visualization complete")
```

## Customer Analysis

```python3
# Customer insights
customer_stats = df.groupby('Customer_ID').agg({
    'Revenue': 'sum',
    'Order_ID': 'count',
    'Rating': 'mean'
}).round(2)

customer_stats.columns = ['Total_Spent', 'Orders', 'Avg_Rating']

print("\nCUSTOMER ANALYSIS:")
print(f"Total Customers: {len(customer_stats)}")
print(f"Average Customer Value: Rs {customer_stats['Total_Spent'].mean():,.2f}")
print(f"Average Orders per Customer: {customer_stats['Orders'].mean():.2f}")
print(f"\nTop 5 Customers by Value:")
print(customer_stats.sort_values('Total_Spent', ascending=False).head())
```

```python3
# Visualize customer analysis
fig, axes = plt.subplots(2, 2, figsize=(15, 12))
fig.suptitle('Customer Analysis', fontsize=16, fontweight='bold', y=1.00)

# Customer value distribution
axes[0, 0].hist(customer_stats['Total_Spent'], bins=15, color='steelblue', edgecolor='black')
axes[0, 0].set_title('Customer Lifetime Value Distribution', fontweight='bold', fontsize=12)
axes[0, 0].set_xlabel('Total Spent (Rs)')
axes[0, 0].set_ylabel('Number of Customers')
axes[0, 0].axvline(customer_stats['Total_Spent'].mean(), color='red', linestyle='--', 
                   linewidth=2, label=f"Mean: Rs {customer_stats['Total_Spent'].mean():,.0f}")
axes[0, 0].legend()

# Orders per customer
axes[0, 1].hist(customer_stats['Orders'], bins=range(1, int(customer_stats['Orders'].max())+2), 
               color='coral', edgecolor='black')
axes[0, 1].set_title('Orders per Customer Distribution', fontweight='bold', fontsize=12)
axes[0, 1].set_xlabel('Number of Orders')
axes[0, 1].set_ylabel('Number of Customers')

# Customer segments
segments = pd.cut(customer_stats['Total_Spent'], 
                  bins=[0, 5000, 10000, 15000, float('inf')],
                  labels=['Low Value', 'Medium Value', 'High Value', 'Premium'])
segment_counts = segments.value_counts()
colors_seg = ['#FFB703', '#FB8500', '#8ECAE6', '#2E86AB']
axes[1, 0].pie(segment_counts, labels=segment_counts.index, autopct='%1.1f%%', 
              colors=colors_seg, startangle=90)
axes[1, 0].set_title('Customer Segmentation by Value', fontweight='bold', fontsize=12)

# Rating vs Spending
axes[1, 1].scatter(customer_stats['Total_Spent'], customer_stats['Avg_Rating'], 
                  alpha=0.6, s=100, color='steelblue', edgecolor='black')
axes[1, 1].set_title('Customer Satisfaction vs Spending', fontweight='bold', fontsize=12)
axes[1, 1].set_xlabel('Total Spent (Rs)')
axes[1, 1].set_ylabel('Average Rating')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
print("Customer analysis visualization complete")
```

## Price and Rating Analysis

```python3
print("\nPRICE ANALYSIS:")
print(f"Minimum Price: Rs {df['Price'].min():,.0f}")
print(f"Maximum Price: Rs {df['Price'].max():,.0f}")
print(f"Average Price: Rs {df['Price'].mean():,.0f}")
print(f"Median Price: Rs {df['Price'].median():,.0f}")

print("\nRATING ANALYSIS:")
print(f"Minimum Rating: {df['Rating'].min()}")
print(f"Maximum Rating: {df['Rating'].max()}")
print(f"Average Rating: {df['Rating'].mean():.2f}")
print(f"Median Rating: {df['Rating'].median():.2f}")
```

```python3
# Visualize price and rating
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
fig.suptitle('Price and Rating Analysis', fontsize=16, fontweight='bold')

# Price distribution
axes[0].hist(df['Price'], bins=20, color='steelblue', edgecolor='black')
axes[0].axvline(df['Price'].mean(), color='red', linestyle='--', linewidth=2, 
               label=f"Mean: Rs {df['Price'].mean():,.0f}")
axes[0].set_title('Price Distribution', fontweight='bold', fontsize=12)
axes[0].set_xlabel('Price (Rs)')
axes[0].set_ylabel('Frequency')
axes[0].legend()
axes[0].grid(True, alpha=0.3, axis='y')

# Rating distribution
rating_counts = df['Rating'].value_counts().sort_index()
axes[1].bar(rating_counts.index, rating_counts.values, color='coral', edgecolor='black', width=0.2)
axes[1].set_title('Customer Rating Distribution', fontweight='bold', fontsize=12)
axes[1].set_xlabel('Rating')
axes[1].set_ylabel('Number of Orders')
axes[1].set_xticks(rating_counts.index)
axes[1].grid(True, alpha=0.3, axis='y')
for i, v in enumerate(rating_counts.values):
    axes[1].text(rating_counts.index[i], v, f'{v}', ha='center', va='bottom', fontsize=9)

plt.tight_layout()
plt.show()
print("Price and rating analysis visualization complete")
```

## Key Insights and Recommendations

```python3
print("\n" + "="*70)
print("KEY INSIGHTS AND BUSINESS RECOMMENDATIONS")
print("="*70)

# Find top performers
top_category = df.groupby('Category')['Revenue'].sum().idxmax()
top_category_revenue = df.groupby('Category')['Revenue'].sum().max()
top_region = df.groupby('Region')['Revenue'].sum().idxmax()
top_region_revenue = df.groupby('Region')['Revenue'].sum().max()
top_product = df.groupby('Product')['Revenue'].sum().idxmax()
top_product_revenue = df.groupby('Product')['Revenue'].sum().max()

print(f"\n1. TOP PERFORMING CATEGORY:")
print(f"   Category: {top_category}")
print(f"   Revenue: Rs {top_category_revenue:,.2f}")
print(f"   Recommendation: Focus marketing efforts on {top_category} category")

print(f"\n2. TOP PERFORMING REGION:")
print(f"   Region: {top_region}")
print(f"   Revenue: Rs {top_region_revenue:,.2f}")
print(f"   Recommendation: Expand operations in {top_region} region")

print(f"\n3. BEST SELLING PRODUCT:")
print(f"   Product: {top_product}")
print(f"   Revenue: Rs {top_product_revenue:,.2f}")
print(f"   Recommendation: Increase inventory of {top_product}")

# Customer loyalty
repeat_customers = (customer_stats['Orders'] > 1).sum()
repeat_rate = (repeat_customers / len(customer_stats)) * 100
print(f"\n4. CUSTOMER LOYALTY:")
print(f"   Repeat Customers: {repeat_customers} ({repeat_rate:.1f}% of customer base)")
print(f"   Recommendation: Implement loyalty program to increase retention")

# Profitability
high_margin_products = df.nlargest(3, 'Profit_Margin')[['Product', 'Profit_Margin']].drop_duplicates()
print(f"\n5. PROFITABILITY:")
print(f"   Average Profit Margin: {avg_profit_margin:.2f}%")
print(f"   Total Profit: Rs {total_profit:,.2f}")
print(f"   Recommendation: Focus on high-margin products")

# Rating insight
print(f"\n6. CUSTOMER SATISFACTION:")
print(f"   Average Rating: {avg_rating:.2f}/5.0")
print(f"   Recommendation: Maintain high quality standards")

print("\n" + "="*70)
```
