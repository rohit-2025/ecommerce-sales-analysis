# E-Commerce Sales Analysis - Data Exploration

## Objectives:
- Load the raw sales data
- Understand the structure of the dataset
- Identify data types and potential issues
- Get initial insights

## Dataset Columns:
- **Order_ID**: Unique order identifier
- **Date**: Transaction date
- **Product**: Product name
- **Category**: Product category
- **Price**: Unit price
- **Quantity**: Units purchased
- **Revenue**: Total revenue (Price × Quantity)
- **Customer_ID**: Customer identifier
- **Region**: Geographic region
- **Rating**: Customer satisfaction rating (1-5)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set display options
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
sns.set_style('whitegrid')

print("Libraries loaded successfully!")
```

## Step 1: Load the Data

```python
# Load CSV file
df = pd.read_csv('../data/raw/sales_data.csv')
print(f"Dataset shape: {df.shape}")
print(f"\nFirst few rows:")
print(df.head(10))
```

## Step 2: Data Overview

```python
# Data types
print("Data Types:")
print(df.dtypes)
print("\n" + "="*50)

# Info about the dataset
print("\nDataset Info:")
df.info()
```

## Step 3: Missing Values Check

```python
# Check for missing values
missing_values = df.isnull().sum()
print("Missing Values:")
print(missing_values)
print(f"\nTotal missing values: {df.isnull().sum().sum()}")
```

## Step 4: Basic Statistics

```python
# Numerical columns statistics
print("Statistical Summary:")
print(df.describe())
```

## Step 5: Categorical Data Exploration

```python
# Unique values in categorical columns
print("Unique Categories:")
print(f"Unique Products: {df['Product'].nunique()}")
print(f"Unique Categories: {df['Category'].nunique()}")
print(f"Regions: {df['Region'].unique()}")
print(f"Unique Customers: {df['Customer_ID'].nunique()}")

print("\nCategory Distribution:")
print(df['Category'].value_counts())
```

## Step 6: Date Analysis

```python
# Convert date to datetime
df['Date'] = pd.to_datetime(df['Date'])

print("Date Range:")
print(f"Start Date: {df['Date'].min()}")
print(f"End Date: {df['Date'].max()}")
print(f"Duration: {(df['Date'].max() - df['Date'].min()).days} days")
```

## Step 7: Key Insights from Exploration

```python
# Summary insights
print("KEY INSIGHTS:")
print("="*50)
print(f"Total Orders: {len(df):,}")
print(f"Total Revenue: ₹{df['Revenue'].sum():,.2f}")
print(f"Average Order Value: ₹{df['Revenue'].mean():,.2f}")
print(f"Total Items Sold: {df['Quantity'].sum():,}")
print(f"Average Rating: {df['Rating'].mean():.2f}/5.0")
print(f"Price Range: ₹{df['Price'].min():.0f} - ₹{df['Price'].max():,.0f}")
print(f"\nTop Category by Orders: {df['Category'].value_counts().index[0]}")
print(f"Top Region: {df['Region'].value_counts().index[0]}")
```

## Step 8: Quick Visualizations

```python
# Create basic visualizations
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Category distribution
df['Category'].value_counts().plot(kind='bar', ax=axes[0, 0], color='steelblue')
axes[0, 0].set_title('Orders by Category', fontsize=12, fontweight='bold')
axes[0, 0].set_ylabel('Number of Orders')
axes[0, 0].tick_params(axis='x', rotation=45)

# 2. Region distribution
df['Region'].value_counts().plot(kind='bar', ax=axes[0, 1], color='coral')
axes[0, 1].set_title('Orders by Region', fontsize=12, fontweight='bold')
axes[0, 1].set_ylabel('Number of Orders')
axes[0, 1].tick_params(axis='x', rotation=0)

# 3. Price distribution
axes[1, 0].hist(df['Price'], bins=20, color='lightgreen', edgecolor='black')
axes[1, 0].set_title('Price Distribution', fontsize=12, fontweight='bold')
axes[1, 0].set_xlabel('Price (₹)')
axes[1, 0].set_ylabel('Frequency')

# 4. Rating distribution
df['Rating'].value_counts().sort_index().plot(kind='bar', ax=axes[1, 1], color='gold')
axes[1, 1].set_title('Rating Distribution', fontsize=12, fontweight='bold')
axes[1, 1].set_ylabel('Number of Orders')
axes[1, 1].set_xlabel('Rating')
axes[1, 1].tick_params(axis='x', rotation=0)

plt.tight_layout()
plt.show()
print("✅ Visualizations created!")
```

## Step 9: Detailed Column Analysis

```python
# Analyze each column in detail
print("\n" + "="*70)
print("DETAILED COLUMN ANALYSIS")
print("="*70)

print("\n1️⃣ ORDER_ID:")
print(f"   Type: {df['Order_ID'].dtype}")
print(f"   Unique Orders: {df['Order_ID'].nunique()}")
print(f"   Duplicates: {df['Order_ID'].duplicated().sum()}")

print("\n2️⃣ DATE:")
print(f"   Type: {df['Date'].dtype}")
print(f"   Date Range: {df['Date'].min()} to {df['Date'].max()}")

print("\n3️⃣ PRICE:")
print(f"   Min: ₹{df['Price'].min():,.0f}")
print(f"   Max: ₹{df['Price'].max():,.0f}")
print(f"   Mean: ₹{df['Price'].mean():,.0f}")
print(f"   Median: ₹{df['Price'].median():,.0f}")

print("\n4️⃣ QUANTITY:")
print(f"   Min: {df['Quantity'].min()}")
print(f"   Max: {df['Quantity'].max()}")
print(f"   Mean: {df['Quantity'].mean():.2f}")

print("\n5️⃣ REVENUE:")
print(f"   Total: ₹{df['Revenue'].sum():,.2f}")
print(f"   Min: ₹{df['Revenue'].min():,.2f}")
print(f"   Max: ₹{df['Revenue'].max():,.2f}")
print(f"   Mean: ₹{df['Revenue'].mean():,.2f}")

print("\n6️⃣ RATING:")
print(f"   Min: {df['Rating'].min()}")
print(f"   Max: {df['Rating'].max()}")
print(f"   Mean: {df['Rating'].mean():.2f}")
print(f"   Median: {df['Rating'].median():.2f}")
```
